# Synthetic Data Generation Notebook

This notebook demonstrates how to generate synthetic data for various use cases using Python. It includes code for defining data generation logic, handling edge cases, and visualizing or exporting the results. The notebook is intended for experimentation and prototyping of synthetic data workflows.

```python
# imports

import os
import json
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr
```

## Import Libraries and Setup
This code block initializes the environment and sets up any required imports or variables for the synthetic data generation process. 

## Load API Key and Initialize OpenAI Client
This code block loads the OpenAI API key from the environment and initializes the OpenAI client for use in data generation.

```python
# Initialization

load_dotenv(override=True)

openai_api_key = os.getenv('OPENAI_API_KEY')
if openai_api_key:
    print(f"OpenAI API Key exists and begins {openai_api_key[:8]}")
else:
    print("OpenAI API Key not set")
    
MODEL = "gpt-4o-mini"
openai = OpenAI()
```

## Define Synthetic Data Generation Function
This code block defines the main function for generating synthetic data using the OpenAI API, including support for edge cases and formatting.

```python
def generate_synth_data(use_case, examples, num_records, edge_cases):
    prompt = f"""
    Generate {num_records} synthetic data records for the following use case:
    {use_case}

    Examples:
    {examples}

    The data should be in JSON format.
    """

    if edge_cases:
        prompt += "\n\nPlease include edge cases and invalid data in the generated data. For example negative numbers or incorrect formats."

    response = openai.chat.completions.create(
        model=MODEL,
        messages=[
            {"role": "system", "content": "You are a helpful assistant that generates synthetic data."},
            {"role": "user", "content": prompt}
        ]
    )

    data = response.choices[0].message.content
    print(data)
    return data
```

## Example: Generate Synthetic Customer Support Data
This code block demonstrates how to use the data generation function to create synthetic customer support ticket data.

```python
generate_synth_data(
    use_case="Customer support ticket classification",
    examples=[
        {"text": "I need help with my order", "label": "order"},
        {"text": "How do I reset my password?", "label": "account"},
        {"text": "I want to return an item", "label": "return"}
    ],
    num_records=10,
    edge_cases=False
)
```

## Gradio Interface for Interactive Data Generation
This code block creates a Gradio web interface to allow users to interactively generate synthetic data for their own use cases.

```python
demo = gr.Interface(
    fn=generate_synth_data,
    inputs=[
        gr.Textbox(label="Use Case"),
        gr.Textbox(label="Examples"),
        gr.Number(label="Number of Records", value=10),
        gr.Checkbox(label="Generate edge-cases and invalid data")
    ],
    outputs=gr.Textbox(label="Synthetic Data"),
    title="Synthetic Data Generator",
    description="Generate synthetic data based on the provided use case and examples.",
    flagging_mode="never"
)
```

```python
demo.launch(share=False)
```
